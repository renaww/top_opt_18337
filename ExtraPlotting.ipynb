```julia
using Plots

np = [35, 126, 737, 4436, 14776]

adjoint_1iter = [
    0.9434222,
    0.0312459,
    0.0797287,
    0.2343847,
    38.1619728
]

fd_1iter = [
    0.6413823,
    2.0185761,
    20.8235929,
    734.938111,
    171 * 60 + 52.9
]

p = plot(
    np,
    adjoint_1iter,
    marker = :circle,
    linewidth = 2,
    label = "Adjoint",
    xlabel = "Number of design variables, np",
    ylabel = "Wall-clock time per MMA iteration (s)",
    title = "Runtime for 1 Iteration of MMA, β = 8, η = 0.5",
    xscale = :log10,
    yscale = :log10,
    legend = :topleft,
    figsize = (1600, 1200)
)

plot!(
    p,
    np,
    fd_1iter,
    marker = :square,
    linewidth = 2,
    label = "Finite Difference",
    figsize = (1600, 1200)
)

display(p)
savefig(p, "one_iteration_runtime_loglog.png")
```

```julia
# -----------------------------------------
# Data: full 500-iteration MMA adjoint only
# -----------------------------------------

adjoint_500iter = [
    19.6,
    26.3,
    39.2,
    3 * 60 + 6.7,
    5 * 60 + 40
]

p2 = plot(
    np,
    adjoint_500iter,
    marker = :circle,
    linewidth = 2,
    label = "Adjoint, 500 iterations",
    xlabel = "Number of design variables, np",
    ylabel = "Total wall-clock time for 500 MMA iterations (s)",
    title = "Full 500-Iteration MMA Runtime, Adjoint Only",
    xscale = :log10,
    yscale = :log10,
    legend = :topleft
)

display(p2)
savefig(p2, "full_500_iteration_runtime_adjoint.png")
```
